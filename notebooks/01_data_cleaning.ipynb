```python
def clean_old_format(sheet_name):
    df = pd.read_excel('../data/raw/manual_spreadsheet/Copia de vendas sorvete.xlsx', sheet_name=sheet_name, header= None)
    header_row = df[df[1] == 'Vendas'].index[0]
    df = pd.read_excel('../data/raw/manual_spreadsheet/Copia de vendas sorvete.xlsx', sheet_name=sheet_name, header=header_row)
    df['Unnamed: 0'] = pd.to_numeric(df['Unnamed: 0'], errors='coerce')
    df.dropna(subset=['Unnamed: 0'], inplace=True)
    df.rename(columns={'Unnamed: 0': 'day', 'Vendas': 'sales', 'Gastos': 'expenses', 'Unnamed: 3': 'expense_category'}, inplace=True)
    df= df[['day', 'sales', 'expenses', 'expense_category']]
    df['expenses'] = pd.to_numeric(df['expenses'], errors='coerce')
    df['month_sheet'] = sheet_name
    return df
```

```python
def clean_new_format(sheet_name):
    df = pd.read_excel('../data/raw/manual_spreadsheet/Copia de vendas sorvete.xlsx', sheet_name=sheet_name, header=None)
    header_row = df[df[0] == 'Data'].index[0]
    df = pd.read_excel('../data/raw/manual_spreadsheet/Copia de vendas sorvete.xlsx', sheet_name=sheet_name, header=header_row)
    df.columns = df.columns.str.strip()
    df.rename(columns={'Data': 'date', 'Vendas': 'sales', 'Gastos': 'expenses', 'Tipo gasto': 'expense_category'}, inplace=True)
    df= df[['date', 'sales', 'expenses', 'expense_category']]
    df['expenses'] = pd.to_numeric(df['expenses'], errors='coerce')
    df['date'] = pd.to_datetime(df['date'], format='%d/%m/%Y', errors='coerce')
    df.dropna(subset=['date'], inplace=True)
    df['month_sheet'] = sheet_name
    return df
```

```python
def clean_kyte_sales(filepath):
    sales_df = pd.read_csv(filepath)
    sales_detail = sales_df[['Date/Time', 'Total', 'Payment Method']]
    sales_detail.rename(columns={'Date/Time': 'sale_time', 'Total': 'amount_sold', 'Payment Method': 'payment_type'}, inplace=True)
    sales_detail['sale_time'] = pd.to_datetime(sales_detail['sale_time'], format='%m/%d/%Y %I:%M %p', errors='coerce')
    sales_detail.loc[sales_detail['payment_type'].str.contains(',', regex=False), 'payment_type'] = 'Multiple'
    
    sales_detail['payment_type'] = sales_detail['payment_type'].replace(payment_mapping)
    return sales_detail
    
    
```
